```python
import nltk
import random
from nltk.chat.util import Chat, reflections

nltk.download('punkt', quiet=True)
nltk.download('wordnet', quiet=True)

# Each tuple: (regex_pattern, [list of possible responses])
# The chatbot picks a random response from the list when a pattern matches.
pairs = [
    (r'hi|hello|hey',
     ['Hello! How can I help you today?',
      'Hi there! What can I do for you?']),

    (r'what is your name\??',
     ['I am a simple rule-based chatbot built with NLTK.',
      'You can call me NLTKBot!']),

    (r'how are you\??',
     ['I am just a program, but I am doing great!',
      'Running perfectly, thanks for asking!']),

    (r'what can you do\??',
     ['I can answer simple questions and have basic conversations.',
      'I match your input to predefined patterns and reply accordingly.']),

    (r'what is (.*) in cloud computing\??',
     ['%1 is an important concept in cloud computing. Could you be more specific?',
      'Great question! %1 relates to scalability and on-demand services.']),

    (r'bye|goodbye|quit',
     ['Goodbye! Have a great day!',
      'See you later!']),

    (r'(.*)',
     ["I'm not sure I understand. Can you rephrase?",
      'Interesting! Tell me more.',
      'Could you clarify what you mean?'])
]

chatbot = Chat(pairs, reflections)
print('Chatbot ready. Type "bye" to exit.')
```

```python
# Interactive loop – type 'bye' to stop
def run_chatbot():
    print('Bot: Hello! I am your cloud computing assistant. Type "bye" to exit.\n')
    while True:
        user_input = input('You: ').strip().lower()
        if not user_input:
            continue
        response = chatbot.respond(user_input)
        print(f'Bot: {response}\n')
        if any(word in user_input for word in ['bye', 'goodbye', 'quit']):
            break

run_chatbot()
```

```python
test_inputs = [
    'hello',
    'what is your name?',
    'what is microservices in cloud computing?',
    'bye'
]

for inp in test_inputs:
    resp = chatbot.respond(inp)
    print(f'You : {inp}')
    print(f'Bot : {resp}\n')
```

```python
import nltk
import re
from nltk.chat.util import Chat, reflections

# Patterns based on our search engine project from tirgul 8
project_pairs = [
    # Greetings
    (r'hi|hello|hey|shalom',
     ['Hello! I am the search engine assistant. Ask me about documents, queries, or services.',
      'Hi! How can I help you with the search engine today?']),

    # About the system
    (r'what (is|are) (this|the) (system|project|search engine)\??',
     ['This is a microservices-based search engine built in Tutorial 8. '
      'It has 4 services: IndexService, QueryService, ResultService, and UserService.']),

    # IndexService
    (r'(.*)(index|indexservice|indexing)(.*)',
     ['IndexService manages all documents and builds an inverted index. '
      'It maps each word to the set of documents containing it.',
      'The IndexService stores documents and allows word-level search via an inverted index.']),

    # QueryService
    (r'(.*)(query|queryservice|search)(.*)',
     ['QueryService receives search terms and supports AND / OR operators. '
      'AND returns documents containing ALL terms; OR returns documents with ANY term.',
      'Use QueryService to search. Example: terms=["cloud", "python"], operator="AND"']),

    # ResultService / ranking
    (r'(.*)(result|ranking|score|resultservice)(.*)',
     ['ResultService formats and ranks results. Ranking is based on term frequency – '
      'documents mentioning your search terms more often get a higher score.',
      'Results are sorted by score (highest first). The score = total occurrences of search terms in the document.']),

    # UserService
    (r'(.*)(user|userservice)(.*)',
     ['UserService manages user data. In production it would connect to a database. '
      'Currently it stores users in memory for demo purposes.']),

    # How to add a document
    (r'how (do i|can i|to) add (a )?document(s)?\??',
     ['Call index_service.add_document({"title": "...", "content": "..."}). '
      'The service automatically builds the inverted index for every word in the content.']),

    # How to search
    (r'how (do i|can i|to) search\??',
     ['Call query_service.create_query({"terms": ["word1", "word2"], "operator": "AND"}). '
      'Then pass the returned query id to result_service.format_results(query_id).']),

    # AND vs OR
    (r'(.*)(and|or|operator)(.*)',
     ['AND = intersection: document must contain ALL search terms. '
      'OR = union: document must contain AT LEAST ONE search term.']),

    # Microservices concept
    (r'(.*)(microservice|architecture)(.*)',
     ['Microservices means each service is independent with a single responsibility. '
      'Our engine has 4 services: Index, Query, Result, User – each can be deployed separately.']),

    # Fallback
    (r'(.*)',
     ["I don't know the answer to that. Try asking about: indexing, querying, ranking, or microservices.",
      'Could you rephrase? I know about: IndexService, QueryService, ResultService, UserService.'])
]

project_chatbot = Chat(project_pairs, reflections)
print('Project chatbot ready!')
```

```python
test_questions = [
    'hello',
    'what is the system?',
    'how does indexservice work?',
    'how do i search?',
    'what is the difference between and or operator?',
    'tell me about ranking',
    'what is the microservice architecture?',
]

for q in test_questions:
    resp = project_chatbot.respond(q)
    print(f'Q: {q}')
    print(f'A: {resp}\n')
```

```python
SYSTEM_CONTEXT = """
You are an assistant for a microservices-based search engine project.
The system has 4 services:
- IndexService: stores documents, builds an inverted index (word -> set of doc_ids)
- QueryService: accepts search terms with AND/OR operators, returns matching doc_ids
- ResultService: formats and ranks results by term frequency score
- UserService: manages users (demo, in-memory)
Answer questions about this project concisely.
"""

def smart_respond(user_input):
    """Try NLTK first; if it gives the generic fallback, use Gemini."""
    nltk_response = project_chatbot.respond(user_input.lower())

    # If NLTK hits the fallback pattern, escalate to Gemini
    fallback_phrases = ["i don't know", "could you rephrase"]
    if any(phrase in nltk_response.lower() for phrase in fallback_phrases):
        prompt = f"{SYSTEM_CONTEXT}\n\nUser question: {user_input}"
        gemini_response = model.generate_content(prompt)
        return f'[Gemini] {gemini_response.text.strip()}'

    return f'[NLTK] {nltk_response}'


# Demo
demo_questions = [
    'how does ranking work in resultservice?',  # NLTK should handle
    'what happens if two documents have the same score?',  # Gemini fallback
    'can i add images as documents?',  # Gemini fallback
]

for q in demo_questions:
    print(f'Q: {q}')
    print(smart_respond(q))
    print()
```
